Logistic Regression predicts from a linear combination of transformed features. C controls regularization. Random Forest averages many trees. Depth and minimum leaf size limit tree complexity. Compare settings using validation ROC-AUC and then F1.

```python
import sys
from pathlib import Path

root = Path.cwd()
if root.name == "notebooks":
    root = root.parent
sys.path.insert(0, str(root))

from src.config import TARGET
from src.ingestion import read_data, clean_data, split_data
from src.preprocessing import fit_preprocessor, transform
from src.traditional_ml import train_traditional
from src.evaluation import classification_metrics, score_models

splits = split_data(clean_data(read_data()))
train, validation = splits["train"], splits["validation"]
preprocessor = fit_preprocessor(train)
x_train, x_validation = transform(preprocessor, train), transform(preprocessor, validation)
models = train_traditional(x_train, train[TARGET], x_validation, validation[TARGET])
for name, scores in score_models(models, x_validation).items():
    print(name, classification_metrics(validation[TARGET], scores))
```
