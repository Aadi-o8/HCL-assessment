Accuracy counts correct classifications. Precision asks how often a churn prediction is correct. Recall asks how much actual churn was detected. F1 balances precision and recall. ROC-AUC measures how well scores rank churn above retention across thresholds. The confusion matrices use a 0.5 threshold.
